```python
import fasttext
import fasttext.util
import numpy as np
```

```python
fasttext.util.download_model('es', if_exists='ignore')
ft_model = fasttext.load_model('cc.es.300.bin')
fasttext.util.reduce_model(ft_model, 100)
```

```python
def get_sentence_embedding(sentence, model):
    words = sentence.split()
    word_vectors = [model.get_word_vector(word) for word in words if word in model]
    if word_vectors:
        return np.mean(word_vectors, axis=0)  
    else:
        return np.zeros(model.get_dimension())  
```

```python
with open("El Alquimista.txt", "r", encoding="utf-8") as file:
    text = file.read()
```

```python
paragraphs = [p.strip() for p in text.split("\n") if p.strip()]
```

```python
embeddings = np.array([get_sentence_embedding(p, ft_model) for p in paragraphs])
```

```python
np.save("el_alquimista_fasttext_embeddings.npy", embeddings)
print("Embeddings guardados en 'el_alquimista_fasttext_embeddings.npy'")
```

```python
print(f"Dimensiones de los embeddings: {embeddings.shape}")  # (N_párrafos, 300)
print("Primer embedding:", embeddings[0][:10])  # Muestra los primeros 10 valores del primer párrafo
```
